# Exploratory figures (Figures 5–7)

Original plotting code for the histogram/KDE plots (Figure 5), box plots (Figure 6), and the Pearson correlation heatmap (Figure 7). Run after the data-preparation cell; set `DATA_DIR` to the folder holding the Kaggle files.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

DATA_DIR = '../data'
cc = pd.read_csv(f'{DATA_DIR}/cc_info.csv')
trans = pd.read_csv(f'{DATA_DIR}/transactions.csv.zip')
df = pd.merge(cc, trans, on='credit_card', how='left')

label_encoder = LabelEncoder()
df['City'] = label_encoder.fit_transform(df['city'])
df['State'] = label_encoder.fit_transform(df['city'])  # as in the original run (Section 3.3)

num_columns = ['credit_card', 'State', 'City', 'zipcode', 'credit_card_limit',
               'transaction_dollar_amount', 'Long', 'Lat']
```

## Figure 5. Histograms with KDE

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

num_columns = ['credit_card', 'State', 'City', 'zipcode', 'credit_card_limit',
               'transaction_dollar_amount', 'Long', 'Lat']

sns.set_style("ticks")

plt.figure(figsize=(15, 7))
for i, col in enumerate(num_columns, 1):
    ax = plt.subplot(2, 4, i)
    # Plot histogram
    sns.histplot(df[col], bins=30,edgecolor='black', color='#9898f8', kde=False, ax=ax, stat='density', alpha=0.5)
    # Overlay density plot
    sns.kdeplot(df[col], color='#4949ff', ax=ax, linewidth=1.5)
    ax.set_title(f'{col}', fontsize=14, fontweight='bold')
    ax.set_xlabel('')  # Remove x-axis label for all
    ax.set_ylabel('Density', fontsize=11, fontweight='bold')
    ax.tick_params(axis='x', labelsize=10)
    ax.tick_params(axis='y', labelsize=10)
    # Bold x-axis tick labels
    for label in ax.get_xticklabels():
        label.set_fontweight('bold')
    for label in ax.get_yticklabels():
        label.set_fontweight('bold')

plt.tight_layout()
plt.show()
```

## Figure 6. Box plots

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

num_columns = ['credit_card', 'State', 'City', 'zipcode', 'credit_card_limit',
               'transaction_dollar_amount', 'Long', 'Lat']

plt.figure(figsize=(15, 6))
for i, col in enumerate(num_columns, 1):
    ax = plt.subplot(2, 4, i)
    sns.boxplot(x=df[col], color='#9898f8', ax=ax)
    ax.set_title(f'{col}', fontsize=14, fontweight='bold')
    ax.set_xlabel('')
    ax.grid(False)
    ax.tick_params(axis='x', labelsize=10)
    ax.tick_params(axis='y', labelsize=10)
    # Bold x-axis tick labels
    for label in ax.get_xticklabels():
        label.set_fontweight('bold')

plt.tight_layout()
plt.show()
```

## Figure 7. Pearson correlation heatmap

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Select only numeric columns
numeric_df = df[num_columns]

# Generate the heatmap using the numeric dataframe with a modified colormap
sns.heatmap(numeric_df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
